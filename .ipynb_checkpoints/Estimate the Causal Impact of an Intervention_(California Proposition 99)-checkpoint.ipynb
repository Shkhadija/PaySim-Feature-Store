```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
print("Ştat sayı:", df["state"].nunique())
print("İllər:", df["year"].min(), "-", df["year"].max())
```

```python
TREATED = "California"
T0 = 1989   # vergi tətbiqinin ilk ili
```

```python
df["treated_unit"] = (df["state"] == TREATED).astype(int)
df["post"] = (df["year"] >= T0).astype(int)
df["D"] = df["treated_unit"] * df["post"]
```

```python
wide = df.pivot(index="year", columns="state", values="cigsale")
print(wide.shape)  
wide.iloc[:5, :5]
```

```python
ca = wide[TREATED]
others_mean = wide.drop(columns=TREATED).mean(axis=1)

plt.figure(figsize=(9, 5))
plt.plot(ca.index, ca, label="California", lw=2)
plt.plot(others_mean.index, others_mean, label="Digər 38 ştatın orta", lw=2)
plt.axvline(1988, color="gray", ls="--", label="Proposition 99")
plt.ylabel("Adambaşına siqaret satışı (paket)")
plt.legend()
plt.title("California vs digər ştatlar")
plt.show()
```

```python
pre = wide.loc[1970:1988]
post = wide.loc[1989:2000]

ca_pre = pre["California"]
oth_pre = pre.drop(columns="California").mean(axis=1)

# 1) Pre-dövrdə fərq (Kaliforniya - digərləri) zamanla necə dəyişir?
gap_pre = ca_pre - oth_pre
print(gap_pre.round(1))

# 2) Pre-dövrdə fərqin trendi (illik dəyişmə)
slope = np.polyfit(gap_pre.index, gap_pre.values, 1)[0]
print("Pre-dövr fərqin illik trendi:", round(slope, 3), "paket/il")

# 3) Sadə DiD
ca_diff = wide.loc[1989:2000, "California"].mean() - ca_pre.mean()
oth_diff = wide.loc[1989:2000].drop(columns="California").mean(axis=1).mean() - oth_pre.mean()
print("DiD təxmini:", round(ca_diff - oth_diff, 2), "paket")
```

```python
# Baza kimi yalnız 1981-1988 götürsək DiD necə dəyişir?
base = wide.loc[1981:1988]
ca_b  = base["California"].mean()
oth_b = base.drop(columns="California").mean(axis=1).mean()

ca_p  = wide.loc[1989:2000, "California"].mean()
oth_p = wide.loc[1989:2000].drop(columns="California").mean(axis=1).mean()

print("DiD (baza 1981-88):", round((ca_p - ca_b) - (oth_p - oth_b), 2))
```

```python
from scipy.optimize import minimize

Y_pre_ca  = wide.loc[1970:1988, "California"].values
donors    = wide.drop(columns="California")
Y_pre_don = donors.loc[1970:1988].values
n = Y_pre_don.shape[1]

# Miqyaslama (rəqəmləri ~1 səviyyəsinə gətirir)
scale = Y_pre_ca.mean()
Yc = Y_pre_ca / scale
Yd = Y_pre_don / scale

def loss(w):
    r = Yc - Yd @ w
    return np.mean(r ** 2)

def grad(w):
    r = Yc - Yd @ w
    return -2 * (Yd.T @ r) / len(Yc)

print("Başlanğıc loss:", loss(np.ones(n) / n))

res = minimize(
    loss,
    x0=np.ones(n) / n,
    jac=grad,
    bounds=[(0, 1)] * n,
    constraints={"type": "eq", "fun": lambda w: w.sum() - 1,
                 "jac": lambda w: np.ones(n)},
    method="SLSQP",
    options={"maxiter": 2000, "ftol": 1e-14},
)

print(res.success, res.message)
print("Son loss:", res.fun)

weights = pd.Series(res.x, index=donors.columns)
print(weights[weights > 0.01].sort_values(ascending=False).round(3))
rmse = np.sqrt(np.mean((Y_pre_ca - Y_pre_don @ res.x) ** 2))
print("Pre-dövr RMSE:", round(rmse, 2))
```

```python
w = res.x
years = wide.index

synth = pd.Series(wide.drop(columns="California").values @ w, index=years)
real  = wide["California"]
gap   = real - synth

# Qrafik 1: real vs sintetik
plt.figure(figsize=(9, 5))
plt.plot(years, real,  label="California (real)", lw=2)
plt.plot(years, synth, label="Synthetic California", lw=2, ls="--")
plt.axvline(1988, color="gray", ls=":", label="Proposition 99")
plt.ylabel("Adambaşına siqaret satışı (paket)")
plt.legend(); plt.title("Real vs Synthetic California")
plt.show()

# Qrafik 2: fərq (təsir)
plt.figure(figsize=(9, 5))
plt.plot(years, gap, lw=2)
plt.axhline(0, color="black", lw=1)
plt.axvline(1988, color="gray", ls=":")
plt.ylabel("Fərq (paket)"); plt.title("Zamanla təsir (real - sintetik)")
plt.show()

# ATT
att = gap.loc[1989:2000].mean()
print("ATT (1989-2000 orta):", round(att, 2), "paket")
print("2000-ci ildə fərq:", round(gap.loc[2000], 2), "paket")
```

```python
def fit_synth(target_state, pre=(1970, 1988)):
    don = wide.drop(columns=target_state)
    y  = wide.loc[pre[0]:pre[1], target_state].values
    X  = don.loc[pre[0]:pre[1]].values
    sc = y.mean()
    yc, Xc = y / sc, X / sc
    m = X.shape[1]

    loss = lambda w: np.mean((yc - Xc @ w) ** 2)
    grad = lambda w: -2 * (Xc.T @ (yc - Xc @ w)) / len(yc)

    r = minimize(loss, np.ones(m) / m, jac=grad,
                 bounds=[(0, 1)] * m,
                 constraints={"type": "eq", "fun": lambda w: w.sum() - 1,
                              "jac": lambda w: np.ones(m)},
                 method="SLSQP", options={"maxiter": 2000, "ftol": 1e-14})

    synth = pd.Series(don.values @ r.x, index=wide.index)
    rmse  = np.sqrt(np.mean((y - X @ r.x) ** 2))
    return wide[target_state] - synth, rmse

gaps, rmses = {}, {}
for st in wide.columns:
    gaps[st], rmses[st] = fit_synth(st)

gaps = pd.DataFrame(gaps)
print("Hazırdır:", gaps.shape)
```

```python
plt.figure(figsize=(10, 6))
for st in gaps.columns:
    if st != "California":
        plt.plot(gaps.index, gaps[st], color="gray", alpha=0.5, lw=1)
plt.plot(gaps.index, gaps["California"], color="blue", lw=2.5, label="California")
plt.axhline(0, color="black", lw=1)
plt.axvline(1988, color="gray", ls=":")
plt.ylabel("Fərq (paket)"); plt.legend()
plt.title("Placebo testi: bütün ştatların fərqləri")
plt.show()

# Post/pre RMSPE nisbəti və p-dəyəri
post_rmspe = np.sqrt((gaps.loc[1989:2000] ** 2).mean())
pre_rmspe  = np.sqrt((gaps.loc[1970:1988] ** 2).mean())
ratio = post_rmspe / pre_rmspe
print(ratio.sort_values(ascending=False).head(5).round(1))
print("Kaliforniyanın sırası:", int(ratio.rank(ascending=False)["California"]), "/", len(ratio))
print("p-dəyəri:", round((ratio >= ratio["California"]).mean(), 3))
```

```python
tab = pd.DataFrame({"pre": pre_rmspe, "post": post_rmspe, "ratio": ratio})
print(tab.sort_values("ratio", ascending=False).head(6).round(2))

# Standart yanaşma: pre-dövr uyğunluğu pis olan ştatları çıxarırıq
ca_pre_rmspe = pre_rmspe["California"]
for k in [5, 2]:
    keep = pre_rmspe[pre_rmspe <= k * ca_pre_rmspe].index
    p = (ratio[keep] >= ratio["California"]).mean()
    print(f"Pre-RMSPE <= {k}x California: {len(keep)} ştat, p = {p:.3f}")

# Qrafik: yalnız yaxşı uyğunlaşanlar (2x)
keep = pre_rmspe[pre_rmspe <= 2 * ca_pre_rmspe].index
plt.figure(figsize=(10, 6))
for st in keep:
    if st != "California":
        plt.plot(gaps.index, gaps[st], color="gray", alpha=0.6, lw=1)
plt.plot(gaps.index, gaps["California"], color="blue", lw=2.5, label="California")
plt.axhline(0, color="black", lw=1); plt.axvline(1988, color="gray", ls=":")
plt.ylabel("Fərq (paket)"); plt.legend()
plt.title("Placebo: pre-dövr uyğunluğu yaxşı olan ştatlar (≤ 2x CA RMSPE)")
plt.show()
```

```python
att_all = gaps.loc[1989:2000].mean()           # hər ştat üçün post-dövr orta fərq
keep = pre_rmspe[pre_rmspe <= 5 * pre_rmspe["California"]].index
att_k = att_all[keep]

print("Kaliforniya ATT:", round(att_all["California"], 2))
print("Ən mənfi 5 ATT:")
print(att_k.sort_values().head(5).round(2))

# İki tərəfli p-dəyəri (|ATT| ən azı Kaliforniya qədər böyük olan ştatların payı)
p_two = (att_k.abs() >= abs(att_all["California"])).mean()
# Birtərəfli (yalnız mənfi istiqamət, çünki vergi satışı azaltmalıdır)
p_one = (att_k <= att_all["California"]).mean()
print(f"p (iki tərəfli): {p_two:.3f}, p (birtərəfli): {p_one:.3f}")

plt.figure(figsize=(8, 4))
plt.hist(att_k.drop("California"), bins=12, color="gray", alpha=0.7)
plt.axvline(att_all["California"], color="blue", lw=2.5, label="California")
plt.xlabel("Post-dövr orta fərq (ATT, paket)"); plt.legend()
plt.title("Placebo ATT paylanması")
plt.show()
```

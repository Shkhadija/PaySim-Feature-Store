### Data Cleaning

CustomerID məlumatı olmayanlar analizdən çıxarıldı. Customer segmentation müştəri səviyyəsində aparıldığı üçün hər transaction-un hansı müştəriyə aid olduğunu bilməyimiz vacibdir.

```python
df[df['InvoiceNo'].astype(str).str.startswith('C')]
```

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
```

```python
df['InvoiceNo'].astype(str).str.startswith('C').sum()
```

InvoiceNo dəyəri `C` ilə başlayan transaction-lar cancellation kimi müəyyən edildi. Bu transaction-lar real satışları əks etdirmədiyi üçün RFM hesablamasına daxil edilmədi.

```python
(df['Quantity'] < 0).sum()
```

```python
df['InvoiceDate'].min(), df['InvoiceDate'].max()
```

```python
(df['UnitPrice'] < 0).sum()
```

```python
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
```

Hər transaction üçün ümumi satış dəyəri `Quantity × UnitPrice` düsturu ilə hesablandı.






```python
snapshot_date = df['InvoiceDate'].max()
snapshot_date
```

RFM-də Recency hesablamaq üçün datasetdəki ən son transaction tarixi snapshot date kimi istifadə edildi. Müştərinin Recency dəyəri onun son alışından snapshot date-ə qədər keçən günlərin sayı ilə müəyyən edildi.

```python
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'TotalPrice': 'sum'
})
```

```python
rfm.columns
```

```python
rfm.columns = ['Recency', 'Frequency', 'Monetary']
```

- **Recency** – müştərinin son alışından neçə gün keçdiyini göstərir.
- **Frequency** – müştərinin neçə fərqli invoice üzrə alış etdiyini göstərir.
- **Monetary** – müştərinin ümumi xərclədiyi məbləği göstərir.

### Checking Feature Skewness

RFM xüsusiyyətlərinin paylanması yoxlanıldı. Frequency və Monetary göstəricilərində yüksək positive skewness müşahidə edildi. Bu, az sayda müştərinin çox yüksək alış tezliyi və xərcləmə dəyərinə sahib olduğunu göstərir.

Bu cür skewed paylanma clustering nəticələrinə təsir göstərə bildiyi üçün növbəti mərhələdə uyğun transformation tətbiq edildi.

```python
# Log transformation

import numpy as np
rfm_log = rfm.copy()

rfm_log['Frequency'] = np.log1p(rfm_log['Frequency'])
rfm_log['Monetary'] = np.log1p(rfm_log['Monetary'])
```

```python
rfm_log.skew()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
rfm_scaled = scaler.fit_transform(rfm_log)
```

```python
rfm_scaled
```

```python
rfm_scaled = pd.DataFrame(
    rfm_scaled,
    index=rfm_log.index,
    columns=rfm_log.columns
)

rfm_scaled.head()
```

Scaling vacibdir, çünki clustering distance-based metodlardan istifadə edir. Features eyni scale-a gətirilməklə heç bir RFM göstəricisinin sadəcə böyük numerical range səbəbindən clustering nəticəsinə üstünlük verməsinin qarşısı alınır.

```python
from sklearn.cluster import KMeans
```

```python
inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

### K-Means: Elbow Method

K-Means üçün uyğun cluster sayını müəyyənləşdirmək məqsədilə `k=2`-dən `k=10`-a qədər müxtəlif cluster sayı test edildi.
Elbow method hər `k` üçün inertia dəyərini müqayisə edir. Inertia cluster daxilindəki variation-u göstərir və cluster sayı artdıqca azalır. Qrafikdə nəzərəçarpacaq dəyişiklikdən sonra azalmanın zəiflədiyi nöqtə uyğun `k` üçün ilkin göstərici kimi istifadə olunur.

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(rfm_scaled)

    score = silhouette_score(rfm_scaled, labels)
    silhouette_scores.append(score)
```

```python
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(rfm_scaled)

score = silhouette_score(rfm_scaled, labels)
score
```

### K-Means: Silhouette Score

Elbow method ilə yanaşı, müxtəlif `k` dəyərlərinin clustering keyfiyyətini müqayisə etmək üçün Silhouette Score istifadə edildi.

Silhouette Score cluster-lərin bir-birindən nə qədər yaxşı ayrıldığını və hər point-in öz cluster-ına nə qədər uyğun olduğunu ölçür. Daha yüksək score daha yaxşı separation və cohesion göstərir.

Test nəticəsində `k=3` ən yüksək Silhouette Score verdi və K-Means üçün 3 cluster seçildi.

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

clusters = kmeans.fit_predict(rfm_scaled)
```

```python
rfm['Cluster'] = clusters
```

```python
cluster_profile = rfm.groupby('Cluster')[['Recency', 'Frequency', 'Monetary']].mean()
cluster_profile
```

###Recency azdırsa → yaxşıdır, çünki müştəri yaxın zamanda alış edib.
###Frequency çoxdursa → yaxşıdır, çünki müştəri daha çox dəfə alış edib.
###Monetary çoxdursa → yaxşıdır, çünki daha çox pul xərcləyib.
## **Cluster 1** - High-value customers
## **Cluster 0** - Mid-Value Customers
Bu müştərilərin alış davranışı orta səviyyədədir.
Onlar aktiv müştərilərdir, lakin Cluster 1 qədər tez-tez alış etmir və yüksək məbləğ xərcləmirlər.
## **Cluster 2** - Low-Value Customers



```python
from sklearn.cluster import AgglomerativeClustering
```

```python
hierarchical_scores = []

for k in range(2, 11):
    hierarchical = AgglomerativeClustering( n_clusters=k )

    labels = hierarchical.fit_predict(rfm_scaled)

    score = silhouette_score(rfm_scaled, labels)
    hierarchical_scores.append(score)
```

```python
plt.plot(range(2, 11), hierarchical_scores, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Hierarchical Clustering - Silhouette Score')
plt.show()
```

```python
hierarchical = AgglomerativeClustering( n_clusters=2)
```

```python
hierarchical_labels = hierarchical.fit_predict(rfm_scaled)

hierarchical_score = silhouette_score( rfm_scaled, hierarchical_labels)

hierarchical_score
```

### Hierarchical Clustering

K-Means nəticəsini müqayisə etmək üçün Agglomerative Hierarchical Clustering tətbiq edildi.

Müxtəlif cluster sayları (`k=2`–`k=10`) test edildi və hər biri üçün Silhouette Score hesablandı. Ən yüksək score `k=2` üçün əldə edildi və bu nəticə Hierarchical Clustering üçün optimal setting kimi qəbul edildi.

```python
from scipy.cluster.hierarchy import linkage, dendrogram

Z = linkage(rfm_scaled, method='ward')

plt.figure(figsize=(12, 6))
dendrogram(Z, truncate_mode='lastp', p=30)

plt.xlabel('Clusters')
plt.ylabel('Distance')
plt.title('Hierarchical Clustering Dendrogram')
plt.show()
```

### Hierarchical Clustering Dendrogram

Dendrogram müştərilərin hierarchical clustering prosesində hansı ardıcıllıqla birləşdirildiyini vizuallaşdırır.
`Ward` method cluster-ləri within-cluster variance-ı mümkün qədər azaldacaq şəkildə birləşdirir.
Dendrogram hierarchy-ni vizuallaşdırmaq üçün istifadə olunur.

```python
from sklearn.cluster import DBSCAN
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

neighbors = NearestNeighbors(n_neighbors=5)
neighbors.fit(rfm_scaled)

distances, indices = neighbors.kneighbors(rfm_scaled)

distances = np.sort(distances[:, 4])

plt.plot(distances)
plt.xlabel('Customers sorted by distance')
plt.ylabel('5th Nearest Neighbor Distance')
plt.title('K-Distance Plot for DBSCAN')
plt.show()
```

### DBSCAN: Selecting `eps`

DBSCAN üçün uyğun `eps` dəyərini müəyyənləşdirmək məqsədilə k-distance plot quruldu.



```python
eps_values = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]

results = []

for eps in eps_values:
    dbscan = DBSCAN(eps=eps)
    labels = dbscan.fit_predict(rfm_scaled)

    n_clusters = len(set(labels) - {-1})
    n_noise = np.sum(labels == -1)

    if n_clusters >= 2:
        mask = labels != -1
        score = silhouette_score(rfm_scaled[mask], labels[mask])
    else:
        score = np.nan

    results.append([eps, n_clusters, n_noise, score])
```

```python
dbscan_results = pd.DataFrame( results, columns=['eps', 'Clusters', 'Noise Points', 'Silhouette Score'])

dbscan_results
```

### DBSCAN Evaluation

DBSCAN müxtəlif `eps` dəyərləri ilə test edildi, test edilmiş DBSCAN setting-ləri arasında ən yüksək uyğun Silhouette Score `0.327` oldu.

DBSCAN-də `-1` label-i heç bir cluster-a aid olmayan noise points-i göstərir.

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)

rfm_pca = pca.fit_transform(rfm_scaled)
```

### PCA Visualization

RFM datasetində 3 scaled feature olduğu üçün məlumatı 2 ölçüdə vizuallaşdırmaq məqsədilə PCA tətbiq edildi.

PCA məlumatı əsas variation-u qoruyaraq daha az ölçülü representation-a çevirir. Burada PCA clustering aparmaq üçün deyil, cluster nəticələrini 2D qrafikdə daha rahat göstərmək üçün istifadə olunur.

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    rfm_pca[:, 0],
    rfm_pca[:, 1],
    c=clusters,
    alpha=0.6
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Customer Segments - K-Means with PCA')
plt.show()
```
